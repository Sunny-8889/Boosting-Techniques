Boosting Techniques

Question 1: What is Boosting in Machine Learning? Explain how it improves weak learners.

Ans - Boosting is an ensemble learning method that combines multiple weak learners, usually small decision trees, into one stronger model. The learners are trained sequentially: each new learner focuses on examples that the previous learners handled poorly. Their predictions are then combined, often using a weighted sum or vote.
A weak learner performs only slightly better than random guessing. Boosting improves it by repeatedly correcting errors, giving difficult examples more attention, and combining the learners’ contributions. This can produce a highly accurate model, though excessive training or insufficient regularization can cause overfitting.


Question 2: What is the difference between AdaBoost and Gradient Boosting in terms of how models are trained?

Ans - AdaBoost trains learners sequentially while increasing the weights of training examples that earlier learners misclassified. Each learner’s contribution to the final prediction depends on its performance; more accurate learners receive greater weight.
Gradient Boosting trains each new learner to reduce the current model’s loss. It fits the learner to the negative gradient of the loss function, which acts like a residual or correction signal. The final model adds each learner’s predictions, typically scaled by a learning rate.
In short, AdaBoost adjusts the importance of training examples, while Gradient Boosting fits new learners to the errors indicated by the chosen loss function.


Question 3: How does regularization help in XGBoost?

Ans - Regularization discourages XGBoost from building an unnecessarily complex model. Its objective includes penalties for the number of leaves and the size of the leaf weights. Parameters such as gamma, reg_alpha (L1), and reg_lambda (L2) control these penalties.
Other controls—including maximum tree depth, minimum child weight, row and feature subsampling, and the learning rate—also limit model complexity. Together, these techniques can reduce overfitting and improve performance on new data.


Question 4: Why is CatBoost considered efficient for handling categorical data?

Ans - CatBoost can work directly with categorical features when their column names or indices are supplied through cat_features. It transforms categories into numerical representations using statistics calculated from the training data. Its ordered boosting and ordered categorical-statistics methods are designed to reduce target leakage and prediction shift.
This can reduce the need for manual one-hot encoding, especially when features have many categories. Missing-value handling and categorical processing details should still be checked for the dataset and CatBoost version being used.


Question 5: What are some real-world applications where boosting techniques are preferred over bagging methods?

Ans - Boosting is often useful when predictive accuracy is a priority and the data contains complex patterns that successive error correction can capture. Examples include:
	Credit scoring, loan default prediction,and fraud detection
  Customer churn prediction and marketing response modeling
  Medical risk classification and diagnosis support
  Sales, demand, and price prediction
  Ranking and recommendation systems
  Predictive maintenance and equipment failure prediction
Bagging methods such as Random Forests can be attractive when robustness and lower variance are priorities, or when parallel training is important. The better approach depends on the data, evaluation objective, and operational constraints.


```python
#Question 6: Write a Python program to:
#Train an AdaBoost Classifier on the Breast Cancer dataset
#Print the model accuracy

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import accuracy_score

# Load the dataset
X, y = load_breast_cancer(return_X_y=True)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Train the AdaBoost classifier
model = AdaBoostClassifier(
    n_estimators=100,
    learning_rate=0.5,
    random_state=42
)
model.fit(X_train, y_train)

# Evaluate on the test set
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Test accuracy: {accuracy:.4f}")
```

```python
#Question 7: Write a Python program to:
  #Train a Gradient Boosting Regressor on the California Housing dataset
  #Evaluate performance using R-squared score

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score

# Load the dataset
X, y = fetch_california_housing(return_X_y=True)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Train the regressor
model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)
model.fit(X_train, y_train)

# Evaluate on the test set
y_pred = model.predict(X_test)
r2 = r2_score(y_test, y_pred)

print(f"Test R-squared: {r2:.4f}")
```

```python
#Question 8: Write a Python program to:
 #Train an XGBoost Classifier on the Breast Cancer dataset
 #Tune the learning rate using GridSearchCV
 #Print the best parameters and accuracy

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier

# Load and split the dataset
X, y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Define the classifier
model = XGBClassifier(
    n_estimators=200,
    max_depth=3,
    subsample=0.9,
    colsample_bytree=0.9,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=1
)

# Tune learning rate using cross-validation
search = GridSearchCV(
    estimator=model,
    param_grid={"learning_rate": [0.01, 0.05, 0.1, 0.2]},
    scoring="accuracy",
    cv=5,
    n_jobs=-1
)
search.fit(X_train, y_train)

# Evaluate the best model on the held-out test set
best_model = search.best_estimator_
y_pred = best_model.predict(X_test)

print("Best parameters:", search.best_params_)
print(f"Test accuracy: {accuracy_score(y_test, y_pred):.4f}")
```

```python
#Question 9: Write a Python program to:
  #Train a CatBoost Classifier
  #Plot the confusion matrix using seaborn


import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score
from catboost import CatBoostClassifier

# Load and split the dataset
X, y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Train the CatBoost classifier
model = CatBoostClassifier(
    iterations=300,
    learning_rate=0.05,
    depth=6,
    loss_function="Logloss",
    verbose=False,
    random_seed=42
)
model.fit(X_train, y_train)

# Predict and calculate the confusion matrix
y_pred = model.predict(X_test).ravel()
cm = confusion_matrix(y_test, y_pred)

print(f"Test accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("Confusion matrix:")
print(cm)

# Plot the confusion matrix
labels = load_breast_cancer().target_names
plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=labels,
    yticklabels=labels
)
plt.xlabel("Predicted label")
plt.ylabel("Actual label")
plt.title("CatBoost Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from catboost import CatBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    average_precision_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# Replace with your dataset path and actual target column
df = pd.read_csv("loan_data.csv")
target_column = "default"

X = df.drop(columns=[target_column])
y = df[target_column].astype(int)

# Identify categorical columns for CatBoost
categorical_columns = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Split data, keeping the class proportions similar
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# CatBoost accepts categorical values as strings; mark missing values
# explicitly so they are handled consistently.
for column in categorical_columns:
    X_train[column] = X_train[column].fillna("Unknown").astype(str)
    X_test[column] = X_test[column].fillna("Unknown").astype(str)

# Train a baseline class-weighted model
model = CatBoostClassifier(
    iterations=600,
    depth=6,
    learning_rate=0.05,
    loss_function="Logloss",
    auto_class_weights="Balanced",
    random_seed=42,
    verbose=False
)

model.fit(
    X_train,
    y_train,
    cat_features=categorical_columns
)

# Evaluate probabilities and predictions at a 0.5 threshold.
# Choose the final threshold using validation data and business costs.
probabilities = model.predict_proba(X_test)[:, 1]
predictions = (probabilities >= 0.5).astype(int)

print(f"Test PR-AUC: {average_precision_score(y_test, probabilities):.4f}")
print(f"Test ROC-AUC: {roc_auc_score(y_test, probabilities):.4f}")
print("Confusion matrix:")
print(confusion_matrix(y_test, predictions))
print("Classification report:")
print(classification_report(y_test, predictions, zero_division=0))
```
